```python
import pandas as pd
```

```python
df=pd.read_csv('results/full_comparison.csv')
```

```python
# Функция для выделения минимального значения в строке
def highlight_min(row):
    is_min = row == row.min()
    return ['font-weight: bold' if v else '' for v in is_min]
```

```python
(df
 .query("Method != 'Prophet' ")
 .pivot_table(columns=['Type', 'Method'], index=['Group', 'Asset'], values='RMSE', aggfunc='min')
 .style.apply(highlight_min, axis=1).format('{:.2f}')
 #.background_gradient(cmap='RdYlGn_r', axis=1)
 )
```
